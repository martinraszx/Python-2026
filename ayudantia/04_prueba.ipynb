```python
import numpy as np
import matplotlib.pyplot as plt  
```

```python
language = ["Python", "C++", "Java", "JavaScript", "C#"]
preferencia = [45, 20, 15, 12, 8]
print(language,preferencia)
```

```python
plt.figure(figsize=(5,4))
plt.bar(language, preferencia, color=["red","green","yellow","purple","pink"])
plt.title("Lenguajes a usar en preferencia")

plt.show()
```
