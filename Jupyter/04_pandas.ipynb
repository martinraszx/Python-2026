#### **Eliminar los datos nulos del dataset**

Importante -> Guardar el dataset sin nulos dentro de una variable

```python
datos_sin_nulos= datos.dropna() # se eliminaron 14 filas
datos_limpios = datos_sin_nulos
```

```python
datos_limpios
```

#### **Funcion para verificar si un genero de videojuego este presente en la lista**

En el ejemplo a verificar si se encuentra el genero rpg en la lista

```python
# Funcion que filtra por genero RPG
def filtro_rpg(genres):
    return 'RPG' in genres

# Aplicar un filtro para consultar y llamar la funcion filtro_rpg()
datosRPG = datos_limpios[datos_limpios['Genres'].apply(filtro_rpg)] # Creando un nuevo dataframe

# Mostrar el dataset filtrado, incluyendo solo las columnas (Titulo y Rating)
nombresRPG = datosRPG[['Title','Rating']]
nombresRPG.head(20) # que se muestren los primeros 20 videojuegos RPG
```

#### **Promedio del rating de videojuegos Indie**

```python
def filtro_indie(genres):
    return 'Indie' in genres

# Aplicar un filtro para consultar y llamar la funcion filtro_rpg()
datosIndie = datos_limpios[datos_limpios['Genres'].apply(filtro_indie)] # Creando un nuevo dataframe

# Mostrar el dataset filtrado, incluyendo solo las columnas (Titulo y Rating)
nombresIndie = datosIndie[['Title','Rating']]
nombresIndie.head(20) # que se muestren los primeros 20 videojuegos RPG
```

```python
promedioIndie = datosIndie[['Rating']] # promedio de videojuegos RPG
promIndie = promedioIndie.mean() # funcion de promedio de numpy
```

```python
promIndie
```

```python
dataset_promedio = pd.DataFrame({'Genero': ['RPG','Indie'], 'Promedio': [promRPG, promIndie]})
```

```python
plt.pie(dataset_promedio['Promedio'], labels=dataset_promedio['Genero'], autopct='%1.1f')
plt.title("comparacion de ratingf entre rpg e indie")
plt.show()
```
